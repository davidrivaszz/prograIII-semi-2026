```python
class estudiantes:
    def __init__(self,nombre,notas,materias,edad,t_pantalla,t_universidad):
        self.nombre = nombre
        self.notas = notas
        self.materias = materias
        self.t_pantalla = t_pantalla
        self.t_universidad = t_universidad
        self.edad = edad

#MEDIA ARITMETICA
def calcular_media_aritmetica(estudiantes):
    sum_notas = sum(estudiantes.notas)
    total_notas = len(estudiantes.notas)
    resultado = sum_notas/total_notas


    return resultado

#key compara frecuencias y la funcion max por si sla no puede hacerlo
#MODA
def calcular_moda(estudiantes):
    notas = estudiantes.notas
    val_unico = set(notas)
    moda = max(val_unico , key=notas.count)

    return moda

#VARIANZA
def calcular_varianza(estudiantes):
    cantidad_notas = len(estudiantes.notas)
    promedio = sum(estudiantes.notas) / cantidad_notas

    suma_cuadrado = 0
    for elemento in estudiantes.notas:
        resta = elemento - promedio
        suma_cuadrado += resta ** 2

    varianza = suma_cuadrado / cantidad_notas

    return varianza

#RANGO
def calcular_rango(estudiantes):
    rango = max(estudiantes.notas)-min(estudiantes.notas)

    return rango

#MEDIANA
def calcular_mediana(estudiantes):
    notas_ordenadas = sorted(estudiantes.notas)
    cantidad_notas = len(notas_ordenadas)
    mitad = cantidad_notas // 2

    if cantidad_notas % 2 == 0:
        mediana = (notas_ordenadas[mitad - 1] + notas_ordenadas[mitad]) / 2
    else:
        mediana = notas_ordenadas[mitad]

    return mediana

#FRECUENCIA
def calcular_frecuencia(estudiantes):
    notas = estudiantes.notas
    frecuencia = {}
    for nota in notas:
        if nota in frecuencia:
            frecuencia[nota] += 1
        else:
            frecuencia[nota] = 1

    return frecuencia


#aun me faltan  funciones



#para guardar los datos de los estudiantes
list_estudiante = [estudiantes("Sharon Nicole Arguello Torres",[10,10,10,10],
                ["Circuitos Digitales","Estadistica computacional",
                 "Administracion de bases de datos III","Ingles Basico"],21,
                 3,1),
                 estudiantes("Rodrigo Benjamín Baires Flores",[10,10,10,10],
                    ["Circuitos Digitales","Estadistica computacional",
                     "Administracion de bases de datos III","Ingles Basico"],20,1.20,6),
                estudiantes("David Alberto Beltran Rivas",[9,9,10,10],
                     ["Circuitos Digitales","Estadistica computacional",
                        "Administracion de bases de datos III","Ingles Basico"],20,0.25,5),
                estudiantes("Anderson Martin Bonilla Argueta",[10,10,10,10],
                     ["Circuitos Digitales","Estadistica computacional",
                        "Administracion de bases de datos III","Ingles Basico"],19,0.30,5),
                estudiantes("Kevin Manrique Campos Granados",[10,10,10,10],
                     ["Circuitos Digitales","Estadistica computacional",
                         "Administracion de bases de datos III","Ingles Basico"],21,2,7),
                estudiantes("Flor Zulema Campos Orellana",[10,10,10,9],
                     ["Circuitos Digitales","Estadistica computacional",
                      "Administracion de bases de datos III","Ingles Basico"],20,0.45,3),
                estudiantes("Ricardo Jose Carcamo Jimenez",[10,10,10,9],
                       ["Circuitos Digitales","Estadistica computacional",
                        "Administracion de bases de datos III","Ingles Basico"],19,0.15,4,),
                estudiantes("Jefferson Steven Castillo Leiva",[10,10,10,9],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],19,0.30,5),
                estudiantes("Kimberly Michelle Castro Zelaya",[10,10,10,9],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],19,0.15,2),
                estudiantes("Oscar Steven Chicas Huezo",[10,10,10,9],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],22,0.7,13),
                estudiantes("Holman Israel Cisneros Romero",[10,10,10,9],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],25,0.10,5),
                estudiantes("Ramael Ernesto Cordova Colindres",[10,10,10,10],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],19,0.30,9),
                estudiantes("Jose Roberto Diaz Diaz",[10,10,10,10],
                            ["Circuitos Digitales","Estadistica computacional",
                             "Administracion de bases de datos III","Ingles Basico"],20,1.30,8)]


print("<<<<Sistema Estadístico Académico>>>>>")

#utilizare esto en caso de crear el menu de operaciones
operacion = ['Media aritmetica',
#Mediana
'Moda',
'Varianza',
#Desviacion estandar,
#Desviacion tipica ,
'Rango',
#Frecuencia


]


for i, persona in enumerate(list_estudiante, start=1):
    print(f"--- Persona #{i} ---")
    print(f"Nombre:    {persona.nombre}")
    print(f"Edad:      {persona.edad}")
    print("-" * 20)
estudiante=int(input("Selecciona un estudiante: "))-1
est_seleccionado = list_estudiante[estudiante]






diccionario_de_valores = {
   "moda":calcular_moda(est_seleccionado),
    "media aritmetica":calcular_media_aritmetica(est_seleccionado),
    "mediana":calcular_mediana(est_seleccionado),
    "Varianza":calcular_varianza(est_seleccionado),
    "Rango":calcular_rango(est_seleccionado),
    "frecuencia":calcular_frecuencia(est_seleccionado)}


formato = ""
for resultados,valores in diccionario_de_valores.items():
    l =  f"• {resultados}: {valores} \n"
    formato  +=  l
    print(f"• {resultados}: {valores}")
```

Media aritmética: es el promedio normal, sumo todas las notas y divido entre cuántas son.

Mediana: ordeno las notas de menor a mayor y agarro la del centro. Si son pares, saco el promedio de las dos de en medio. Esta no se deja engañar por una nota bien rara (muy alta o muy baja).

Moda: es la nota que más se repite, la que sale más veces en la lista.

Varianza: mide qué tan alejadas están las notas del promedio. Le resto el promedio a cada nota, la elevo al cuadrado pa' que no me den negativos, y saco el promedio de eso. Si todas las notas son iguales me da 0.

Rango: es bien fácil, solo resto la nota más alta menos la más baja, pa' ver qué tan amplio es el rango de notas.

Frecuencia: es un conteo, cuántas veces se repite cada nota. Con esto se ayuda a sacar la moda, porque la moda es la que tiene más frecuencia.

En corto: media, mediana y moda buscan un valor "típico" de las notas, mientras que varianza y rango dicen qué tan dispersas están, y la frecuencia es solo el conteo que respalda todo eso.
