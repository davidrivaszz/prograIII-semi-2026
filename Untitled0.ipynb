```python
fahreheit_celcius = lambda f: (f-32) * 5/9

f = float(input("ingrese los grados fahrenheit:"))

print("grados celcius:", fahreheit_celcius(f))
```

```python
celcius_kelvin = lambda c: c + 273.15

c = float(input("ingrese los grados celcius:"))

print("grados klevin:", celcius_kelvin(c))
```
