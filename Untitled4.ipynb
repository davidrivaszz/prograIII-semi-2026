```python
# ==========================================
# SISTEMA ESTADISTICO ACADEMICO
# ==========================================

import json
import math


# ==========================================
# 1. DATOS EN JSON
# ==========================================

datos_json = '''
[
    {"nombre":"Ana", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":5},
    {"nombre":"Carlos", "materia":"Matematica", "nota":7.5, "edad":20, "traslado":40, "telefono":6},
    {"nombre":"Maria", "materia":"Programacion", "nota":9.0, "edad":19, "traslado":25, "telefono":4},
    {"nombre":"Jose", "materia":"Matematica", "nota":8.0, "edad":21, "traslado":50, "telefono":7},
    {"nombre":"Sofia", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":35, "telefono":5},
    {"nombre":"Luis", "materia":"Matematica", "nota":7.0, "edad":22, "traslado":60, "telefono":8},
    {"nombre":"Daniela", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":6},
    {"nombre":"Pedro", "materia":"Matematica", "nota":7.5, "edad":23, "traslado":45, "telefono":7},
    {"nombre":"Laura", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":20, "telefono":4},
    {"nombre":"Miguel", "materia":"Matematica", "nota":8.0, "edad":21, "traslado":55, "telefono":6},
    {"nombre":"Valeria", "materia":"Programacion", "nota":8.5, "edad":19, "traslado":30, "telefono":5},
    {"nombre":"Andres", "materia":"Matematica", "nota":7.0, "edad":22, "traslado":40, "telefono":8},
    {"nombre":"Gabriela", "materia":"Programacion", "nota":9.0, "edad":20, "traslado":35, "telefono":5},
    {"nombre":"Fernando", "materia":"Matematica", "nota":7.5, "edad":24, "traslado":65, "telefono":7},
    {"nombre":"Paola", "materia":"Programacion", "nota":8.5, "edad":21, "traslado":30, "telefono":4},
    {"nombre":"Ricardo", "materia":"Matematica", "nota":8.0, "edad":22, "traslado":50, "telefono":6},
    {"nombre":"Camila", "materia":"Programacion", "nota":9.0, "edad":19, "traslado":25, "telefono":5},
    {"nombre":"Diego", "materia":"Matematica", "nota":7.5, "edad":23, "traslado":60, "telefono":8},
    {"nombre":"Elena", "materia":"Programacion", "nota":8.0, "edad":20, "traslado":40, "telefono":6},
    {"nombre":"Roberto", "materia":"Matematica", "nota":8.5, "edad":21, "traslado":45, "telefono":7}
]
'''

# Convertimos el JSON a una lista de diccionarios
datos = json.loads(datos_json)


# ==========================================
# 2. CLASE PARA POO
# ==========================================

class Estadistica:

    def __init__(self, datos):
        self.datos = datos

    # --------------------------------------
    # Obtener los valores
    # --------------------------------------

    def obtener_valores(self, campo):

        valores = []

        for estudiante in self.datos:
            valores.append(estudiante[campo])

        return valores


# Creamos el objeto
estadistica = Estadistica(datos)


# ==========================================
# 3. FUNCION MEDIA
# ==========================================

def calcular_media(lista):

    suma = 0

    for numero in lista:
        suma = suma + numero

    media = suma / len(lista)

    return media


# ==========================================
# 4. FUNCION MEDIANA
# ==========================================

def calcular_mediana(lista):

    lista = sorted(lista)

    cantidad = len(lista)

    if cantidad % 2 == 0:

        mitad1 = lista[cantidad // 2 - 1]
        mitad2 = lista[cantidad // 2]

        mediana = (mitad1 + mitad2) / 2

    else:

        mediana = lista[cantidad // 2]

    return mediana


# ==========================================
# 5. FUNCION MODA
# ==========================================

def calcular_moda(lista):

    mayor = 0
    moda = None

    for numero in lista:

        cantidad = lista.count(numero)

        if cantidad > mayor:

            mayor = cantidad
            moda = numero

    if mayor == 1:

        return "No existe moda"

    return moda


# ==========================================
# 6. FUNCION VARIANZA
# ==========================================

def calcular_varianza(lista):

    media = calcular_media(lista)

    suma = 0

    for numero in lista:

        diferencia = numero - media

        suma = suma + diferencia ** 2

    varianza = suma / len(lista)

    return varianza


# ==========================================
# 7. FUNCION DESVIACION ESTANDAR
# ==========================================

def calcular_desviacion_estandar(lista):

    varianza = calcular_varianza(lista)

    desviacion = math.sqrt(varianza)

    return desviacion


# ==========================================
# 8. FUNCION RANGO
# ==========================================

def calcular_rango(lista):

    mayor = max(lista)
    menor = min(lista)

    rango = mayor - menor

    return rango


# ==========================================
# 9. FUNCION FRECUENCIA
# ==========================================

def calcular_frecuencia(lista):

    frecuencias = {}

    for numero in lista:

        if numero in frecuencias:

            frecuencias[numero] = frecuencias[numero] + 1

        else:

            frecuencias[numero] = 1

    return frecuencias


# ==========================================
# 10. MOSTRAR ESTADISTICAS
# ==========================================

def mostrar_estadisticas(campo):

    # Obtener los valores
    lista = estadistica.obtener_valores(campo)

    print("\n================================")
    print("RESULTADOS ESTADISTICOS")
    print("================================")

    print("Datos:", lista)

    print("Media aritmetica:",
          round(calcular_media(lista), 2))

    print("Mediana:",
          round(calcular_mediana(lista), 2))

    print("Moda:",
          calcular_moda(lista))

    print("Varianza:",
          round(calcular_varianza(lista), 2))

    desviacion = calcular_desviacion_estandar(lista)

    print("Desviacion estandar:",
          round(desviacion, 2))

    # Desviacion tipica es igual
    # a la desviacion estandar

    print("Desviacion tipica:",
          round(desviacion, 2))

    print("Rango:",
          round(calcular_rango(lista), 2))

    print("\nFRECUENCIA")

    frecuencias = calcular_frecuencia(lista)

    for valor in sorted(frecuencias):

        print(valor, "->", frecuencias[valor])


# ==========================================
# 11. MOSTRAR LOS ESTUDIANTES
# ==========================================

def mostrar_estudiantes():

    print("\n================================")
    print("LISTA DE ESTUDIANTES")
    print("================================")

    for estudiante in datos:

        print(
            estudiante["nombre"],
            "|", estudiante["materia"],
            "| Nota:", estudiante["nota"],
            "| Edad:", estudiante["edad"],
            "| Traslado:", estudiante["traslado"],
            "min | Telefono:", estudiante["telefono"],
            "horas"
        )


# ==========================================
# 12. MENU PRINCIPAL
# ==========================================

while True:

    print("\n================================")
    print(" SISTEMA ACADEMICO")
    print("================================")

    print("1. Ver estudiantes")
    print("2. Estadisticas de notas")
    print("3. Estadisticas de edades")
    print("4. Estadisticas de traslado")
    print("5. Estadisticas del telefono")
    print("6. Salir")

    opcion = input("Seleccione una opcion: ")

    if opcion == "1":

        mostrar_estudiantes()

    elif opcion == "2":

        mostrar_estadisticas("nota")

    elif opcion == "3":

        mostrar_estadisticas("edad")

    elif opcion == "4":

        mostrar_estadisticas("traslado")

    elif opcion == "5":

        mostrar_estadisticas("telefono")

    elif opcion == "6":

        print("Programa terminado.")
        break

    else:

        print("Opcion incorrecta.")
```

```python
# CONVERSOR DE AREA - EL SALVADOR


# Diccionario: cuantos metros cuadrados equivale UNA unidad de cada tipo
unidades = {
    "a": ("Pie cuadrado",   0.092903),
    "b": ("Vara cuadrada",  0.698896),
    "c": ("Yarda cuadrada", 0.836127),
    "d": ("Metro cuadrado", 1),
    "e": ("Tarea",          698.896),
    "f": ("Manzana",        6988.96),
    "g": ("Hectarea",       10000),
}


def convertir(valor, origen, destino):
    # 1. paso el valor a metros cuadrados
    metros = valor * unidades[origen][1]
    # 2. de metros cuadrados lo paso a la unidad destino
    return metros / unidades[destino][1]


# --- Programa principal ---
print("a) Pie cuadrado   b) Vara cuadrada   c) Yarda cuadrada")
print("d) Metro cuadrado e) Tarea           f) Manzana   g) Hectarea")

origen = input("Unidad que tiene: ")
destino = input("Unidad a la que quiere convertir: ")
valor = float(input("Valor a convertir: "))

resultado = convertir(valor, origen, destino)

print(f"{valor} {unidades[origen][0]} = {round(resultado, 4)} {unidades[destino][0]}")
```
