```python
#MAP
semana = ["domingo", "lunes", "Martes", "Miercoles", "jueves", "Viernes", "sabado"]
print(semana,"Primer dia de la semana: ", semana[0])

primeraLetraDia = map(lambda dia: dia[0], semana)
print(list(primeraLetraDia))

primeraLetraMayusDia = map(lambda dia: dia[0].upper(), semana)
print(list(primeraLetraMayusDia))
```

```python

serie = [2,5,6,8,9,10,12]

pares = filter(lambda num: num%2==0, serie)
impares = filter(lambda num: num%2!=0, serie)
print("Original: ", serie)
print("Pares: ", list(pares))
print("Impares: ", list(impares))
```

```python

notas = [1,6,8,2,7,10,5]

aprobado = 6

aprobados = filter(lambda nota: nota>=aprobado, notas)
reprobados = filter(lambda nota: nota<aprobado, notas)
print("Notas: ", notas)
print("Aprobados: ", list(aprobados))
print("Reprobados: ", list(reprobados))
```

```python

notas = input("Notas: ") #5,6,1,8,9
notas = notas.split(",")
aprobados = filter(lambda nota: float(nota)>=6, notas)
reprobados = filter(lambda nota: float(nota)<6, notas)
print("Notas: ", notas)
print("Aprobados: ", list(aprobados))
print("Reprobados: ", list(reprobados))
```

```python

from functools import reduce

numeros = [1,2,8,4,9,6]
pares = filter(lambda num: num%2==0, numeros)

print("Numeros:" , numeros)
#print("Pares: ", list(pares))

sumaPares = reduce(lambda acum, num: num+acum, pares, 0)
print("Suma de pares: ", sumaPares)
```

```python
import datetime
fecha = datetime.date.today()

dia = fecha.day
mes = fecha.month
year = fecha.year

print(fecha, "Dia: ", dia, "Mes: ", mes, "Año: ", year)

semana = ("domingo", "lunes", "Martes", "Miercoles", "jueves", "Viernes", "sabado")
meses = ("", "Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio", "Julio", "Agosto",
         "Septiembre", "Octubre", "Noviembre", "Diciembre")

print(semana, "Primer dia de la semana: ", semana[0])

diaSemana = semana[fecha.isoweekday()]
print(diaSemana, dia, "de", meses[mes], "del año", year)
```

```python

seguir=True
while(seguir):
  edad = int(input("Edad: "))

  if(edad<0):
    seguir=False
    print("Edad incorrecta")
  elif(edad<=2):
    print("Eres un bebe")
  elif(edad<=11):
    print("Eres un niño")
  elif(edad<18):
    print("Eres un adolescente")
  elif(edad<65):
    print("Eres un adulto")
  elif(edad<=85):
    print("Eres un adulto mayor")
  elif(edad<=130):
    print("Larga vida")
  else:
      print("Edad incorrecta")
      seguir=False
```
